## **1. Setting Up Entrez**

```python
from Bio import Entrez

# Always set your email before using Entrez
Entrez.email = 'your.email@example.com'
```

## **2. Searching NCBI Databases with `esearch()`**

```python
handle = Entrez.esearch(db='nucleotide', term='COX1[Gene] AND Homo sapiens[Organism]', retmax=5)
record = Entrez.read(handle)
handle.close()

print(record['IdList'])
```

**Exercise**: Modify the query to search for sequences related to the BRCA1 gene in *Mus musculus*.

## **3. Fetching Sequence Records with `efetch()`**

```python
ids = record['IdList']
handle = Entrez.efetch(db='nucleotide', id=ids, rettype='gb', retmode='text')
gb_data = handle.read()
handle.close()

print(gb_data[:1000])  # Show a snippet
```

## **4. Parsing GenBank Files with `Bio.SeqIO`**

```python
from Bio import SeqIO
from io import StringIO

gb_handle = StringIO(gb_data)
for seq_record in SeqIO.parse(gb_handle, 'genbank'):
    print(f'ID: {seq_record.id}')
    print(f'Description: {seq_record.description}')
    print(f'Sequence length: {len(seq_record.seq)}')
```

**Exercise**: Extract the gene features from each record and print their names and locations.

## **5. Exploring Other Databases**

```python
handle = Entrez.esearch(db='protein', term='hemoglobin[Title] AND human[Organism]', retmax=3)
record = Entrez.read(handle)
print(record['IdList'])
```
